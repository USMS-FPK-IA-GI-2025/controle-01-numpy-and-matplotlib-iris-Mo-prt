# Partie A — Importation du Dataset, Arrays, Premières Manipulations

### A1. Importation du dataset
Importez NumPy et Matplotlib.  
Chargez le fichier `iris.csv` avec `np.genfromtxt` (`delimiter=','`, `skip_header=1`).  
Stockez les données dans un array nommé `data` et affichez sa `shape`.


```python

import numpy as np
import matplotlib.pyplot as plt


data = np.genfromtxt('/content/Iris.csv', delimiter=',', skip_header=1)
data.shape
```

# **Interprétation :**

Dataset contient 150 échantillons (lignes) et 5 variables (colonnes). Cela correspond bien au dataset Iris classique, avec 4 caractéristiques (longueur/largeur des sépales/pétales) et 1 colonne cible (espèce).

### A2. Séparation des features et des espèces
Séparez :
- les 4 premières colonnes numériques dans un array `features`

```python
# A2. Séparation des features et des espèces
features = data[:, :4]  # Les 4 premières colonnes
species = data[:, 4]    # La dernière colonne

print("Shape de features :", features.shape)
print("Shape de species :", species.shape)
print("\nAperçu de species (valeurs uniques) :", np.unique(species))
```

# Partie B — Statistiques Basiques et Opérations Mathématiques

Calculez la moyenne de chaque colonne de `features` avec `np.mean(axis=0)`.

### B7. Min et Max
Trouvez la valeur minimale et maximale de la longueur des pétales.

Les valeurs extrêmes de la longueur des pétales révèlent une forte variabilité au sein du dataset Iris. Le minimum de 1,00 cm correspond à un iris Setosa, espèce caractérisée par des pétales très courts. Le maximum de 6,90 cm représente un iris Virginica, dont les pétales sont les plus longs.

La normalisation des longueurs des pétales, réalisée par division par la valeur maximale (6,90 cm), a produit un tableau de 150 valeurs comprises entre 0,145 et 1,000. ### C12. Reshape par espèce
Transformez `features` en un array de forme `(3, 50, 4)` nommé `features_by_species`.


```python
# C12. Reshape par espèce
features_by_species = features.reshape(3, 50, 4)

print("features_by_species shape :", features_by_species.shape)
print("Type :", type(features_by_species))
```

Cette réorganisation regroupe les échantillons par espèce : la première dimension (3) correspond aux trois espèces, la deuxième dimension (50) aux 50 échantillons de chaque espèce, et la troisième dimension (4) aux quatre caractéristiques botaniques. ### C13. Concaténation
Concaténez horizontalement les moyennes par colonne avec un array de zéros.


```python

# Moyennes par colonne (calculées précédemment)
moyennes = np.mean(features, axis=0)

# Création d'un array de zéros de même taille
zeros = np.zeros(moyennes.shape)

# Concaténation horizontale
concatenated = np.hstack((moyennes, zeros))

print("Moyennes :", moyennes)
print("Zéros :", zeros)
print("Concaténé :", concatenated)
print("Shape concaténé :", concatenated.shape)
```

**Interprétation :**  La concaténation horizontale a fusionné l'array des moyennes (4 valeurs) avec un array de zéros de même dimension, produisant un nouvel array de 8 éléments. Les 4 premières positions contiennent les moyennes des caractéristiques (~5,84, 3,05, 3,76, 1,20 cm), tandis que les 4 dernières sont initialisées à zéro. Cette opération permet d'étendre un vecteur de caractéristiques avec des valeurs de remplissage, une technique utile pour préparer des données avant certaines analyses ou pour aligner des dimensions dans des opérations matricielles ultérieures.

### C14. Data cleaning (simulation)
Simulez des valeurs NaN et remplacez-les avec `np.nan_to_num`.


```python

features_with_nan = features.copy()
features_with_nan[0, 0] = np.nan
features_with_nan[10, 2] = np.nan
print("Avant nettoyage (avec NaN) :")
print(features_with_nan[:15, :])

features_cleaned = np.nan_to_num(features_with_nan)
print("\nAprès nettoyage (NaN remplacés par 0) :")
print(features_cleaned[:15, :])
```

**Interprétation :**  La simulation de nettoyage de données a inséré deux valeurs NaN dans l'array `features` : à la position (0,0) et (10,2). La fonction `np.nan_to_num()` a remplacé ces NaN par des zéros, comme visible dans l'affichage (0. à la position [0,0] et 0. à la position [10,2]). Cette opération illustre un traitement basique des données manquantes en les remplaçant par une valeur par défaut (zéro), ce qui permet de conserver la structure du dataset tout en éliminant les valeurs non numériques qui pourraient perturber les calculs ultérieurs.

### E24. Bar chart avec error bars
Moyennes par espèce avec écart-type.


```python
especes_labels = ['Setosa', 'Versicolor', 'Virginica']
caract_labels = ['SepalLength', 'SepalWidth', 'PetalLength', 'PetalWidth']

# Calcul des moyennes et écarts-types par espèce
moyennes_par_espece = np.zeros((3, 4))
std_par_espece = np.zeros((3, 4))

for i in range(3):
    start = i * 50
    end = (i + 1) * 50
    moyennes_par_espece[i] = np.mean(features[start:end], axis=0)
    std_par_espece[i] = np.std(features[start:end], axis=0)

# Bar chart avec error bars
x = np.arange(len(especes_labels))
width = 0.2

fig, ax = plt.subplots(figsize=(10, 6))
for i in range(4):
    ax.bar(x + i*width - 1.5*width, moyennes_par_espece[:, i], width,
           yerr=std_par_espece[:, i], capsize=5, label=caract_labels[i])

ax.set_xlabel('Espèce')
ax.set_ylabel('Valeur moyenne (cm)')
ax.set_title('Moyennes des caractéristiques par espèce avec écart-type')
ax.set_xticks(x)
ax.set_xticklabels(especes_labels)
ax.legend()
ax.grid(True, axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

**Interprétation :**  
Le diagramme en barres avec barres d'erreur montre clairement les différences morphologiques entre espèces :

**Tendances générales :**
- Les moyennes augmentent progressivement de Setosa à Virginica pour toutes les caractéristiques sauf SepalWidth
- SepalWidth est maximal chez Setosa (~3,4 cm) et minimal chez Versicolor (~2,8 cm)

**Variabilité (écarts-types) :**

### F29. Variance par espèce
Identifiez l’espèce avec la plus grande variance globale et visualisez ses distributions.


```python
variances = np.var(features_by_species, axis=(1, 2))
especes = ['Setosa', 'Versicolor', 'Virginica']

print("Variance globale par espèce :")
for i in range(3):
    print(f"{especes[i]} : {variances[i]:.4f}")

espece_max_var = especes[np.argmax(variances)]
print(f"\nEspèce avec la plus grande variance : {espece_max_var}")

# Visualisation des distributions pour l'espèce à plus grande variance
indice_max = np.argmax(variances)
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
caract_noms = ['SepalLength', 'SepalWidth', 'PetalLength', 'PetalWidth']

for i, ax in enumerate(axes.flat):
    ax.hist(features_by_species[indice_max, :, i], bins=10, edgecolor='black', alpha=0.7)
    ax.set_title(f'{caract_noms[i]} - {espece_max_var}')
    ax.set_xlabel('cm')
    ax.set_ylabel('Fréquence')
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

Les largeurs (SepalWidth et PetalWidth) présentent également des dispersions significatives. Cette variabilité accrue suggère une plus grande diversité phénotypique au sein de l'espèce Virginica par rapport à Setosa (variance 0,13) et Versicolor (variance 0,24), qui sont plus homogènes. Cette observation pourrait refléter une plasticité morphologique plus importante ou un pool génétique plus diversifié chez Virginica.
